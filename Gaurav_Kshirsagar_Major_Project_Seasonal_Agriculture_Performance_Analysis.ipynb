```python
# Check Missing Values

missing_values = df.isnull().sum()

print(missing_values)
```

```python
missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_percentage.sort_values(ascending=False)
```

```python
# Remove Missing Values

df = df.dropna()

print("New Shape :", df.shape)
```

```python
print("Duplicate Rows Before:", df.duplicated().sum())

df = df.drop_duplicates()

print("Duplicate Rows After:", df.duplicated().sum())
```

```python
df.dtypes
```

```python
numeric_columns = df.select_dtypes(include=['int64', 'float64']).columns

numeric_columns
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

for col in numeric_columns:
    plt.figure(figsize=(6,3))
    sns.boxplot(x=df[col])
    plt.title(f"Boxplot of {col}")
    plt.show()
```

```python
for col in numeric_columns:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    df = df[(df[col] >= lower) & (df[col] <= upper)]

print("Shape after Outlier Removal:", df.shape)
```

```python
plt.figure(figsize=(7,5))
sns.countplot(data=df, x='Season', palette='viridis')
plt.title("Distribution of Farms Across Seasons")
plt.xlabel("Season")
plt.ylabel("Count")
plt.xticks(rotation=0)
plt.show()
```

```python
top_profit = (
    df.groupby('Crop')['Profit_INR']
      .mean()
      .sort_values(ascending=False)
      .head(10)
)

plt.figure(figsize=(10,5))

top_profit.plot(kind='bar')

plt.title("Top 10 Crops by Average Profit")
plt.ylabel("Average Profit (INR)")
plt.xticks(rotation=45)

plt.show()
```

```python
top_state = (
    df.groupby('State')['Profit_INR']
      .mean()
      .sort_values(ascending=False)
      .head(10)
)

plt.figure(figsize=(10,5))

top_state.plot(kind='bar')

plt.title("Top 10 States by Average Profit")
plt.ylabel("Average Profit (INR)")
plt.xticks(rotation=45)

plt.show()
```
